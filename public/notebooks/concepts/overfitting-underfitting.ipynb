# Overfitting and Underfitting

Why a model can look perfect in training and still fail, how to spot both problems, and how to fix them.

This notebook comes from the [Overfitting and Underfitting](https://pabl0aranda.github.io/Atlas-ML/concepts/overfitting-underfitting/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)
```

## Code from the page

```python
from sklearn.model_selection import learning_curve, validation_curve
from sklearn.neighbors import KNeighborsClassifier

ks = [1, 5, 25, 85]
train_scores, val_scores = validation_curve(
    KNeighborsClassifier(), X, y, param_name="n_neighbors", param_range=ks, cv=5,
)
print(train_scores.mean(axis=1), val_scores.mean(axis=1))

sizes, train_scores, val_scores = learning_curve(KNeighborsClassifier(n_neighbors=25), X, y, cv=5)
```
