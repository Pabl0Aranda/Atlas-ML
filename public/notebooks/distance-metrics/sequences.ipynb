## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import numpy as np

# The same wave, once shifted by 10 steps.
t = np.linspace(0, 4 * np.pi, 100)
series_a = np.sin(t)
series_b = np.sin(t - 0.5)
```

## Code from the page

```python
from rapidfuzz.distance import Levenshtein
from dtaidistance import dtw

Levenshtein.distance("kitten", "sitting")  # 3
dtw.distance(series_a, series_b)
```
