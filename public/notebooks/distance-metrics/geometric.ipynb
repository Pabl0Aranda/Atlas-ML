# Euclidean, Manhattan and Chebyshev

The geometric distances, and how Minkowski distance connects them through a single parameter.

This notebook comes from the [Euclidean, Manhattan and Chebyshev](https://pabl0aranda.github.io/Atlas-ML/distance-metrics/geometric/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Code from the page

```python
from scipy.spatial import distance

distance.euclidean([0, 0], [3, 4])  # 5.0
distance.cityblock([0, 0], [3, 4])  # 7
distance.chebyshev([0, 0], [3, 4])  # 4
distance.minkowski([0, 0], [3, 4], p=3)
```
