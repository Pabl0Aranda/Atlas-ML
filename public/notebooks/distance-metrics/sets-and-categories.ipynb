# Hamming and Jaccard

Distances for data that is not numeric, such as categories, binary codes and sets.

This notebook comes from the [Hamming and Jaccard](https://pabl0aranda.github.io/Atlas-ML/distance-metrics/sets-and-categories/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Code from the page

```python
from scipy.spatial import distance

# SciPy returns the share of positions that differ, not the count.
distance.hamming(list("karolin"), list("kathrin"))  # 0.428...

# SciPy's jaccard takes boolean vectors, not sets.
distance.jaccard([1, 1, 1, 0], [0, 1, 1, 1])  # 0.5
```
