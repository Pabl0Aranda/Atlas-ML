# Cosine and Mahalanobis

Distances that look past raw coordinates, to compare direction or to account for the shape of the data.

This notebook comes from the [Cosine and Mahalanobis](https://pabl0aranda.github.io/Atlas-ML/distance-metrics/direction-and-shape/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import numpy as np

# Correlated data, where the Mahalanobis distance differs from the Euclidean one.
rng = np.random.default_rng(7)
X = rng.multivariate_normal([0, 0], [[1, 0.8], [0.8, 1]], size=300)
```

## Code from the page

```python
import numpy as np
from scipy.spatial import distance

distance.cosine([1, 2], [2, 4])  # 0.0

inverse = np.linalg.inv(np.cov(X, rowvar=False))
distance.mahalanobis([0, 0], [0.3, 0.3], inverse)
```
