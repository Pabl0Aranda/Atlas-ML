# Proximal Policy Optimization

Learn a policy directly, while clipping each update so that it cannot change too much at once.

This notebook comes from the [Proximal Policy Optimization](https://pabl0aranda.github.io/Atlas-ML/algorithms/proximal-policy-optimization/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Code from the page

```python
import torch

def ppo_loss(new_log_probs, old_log_probs, advantages, entropy, clip=0.2, entropy_bonus=0.01):
    ratio = torch.exp(new_log_probs - old_log_probs)
    clipped = torch.clamp(ratio, 1 - clip, 1 + clip)
    surrogate = torch.min(ratio * advantages, clipped * advantages)
    return -surrogate.mean() - entropy_bonus * entropy.mean()
```

## Try it

```python
# The loss for a batch of 16 actions with random log-probabilities and advantages.
new_log_probs = torch.randn(16) * 0.1 - 1
old_log_probs = new_log_probs.detach() + torch.randn(16) * 0.1
print(ppo_loss(new_log_probs, old_log_probs, torch.randn(16), torch.rand(16)))
```
