# Gaussian Process Regression

A distribution over functions that gives predictions with honest uncertainty.

This notebook comes from the [Gaussian Process Regression](https://pabl0aranda.github.io/Atlas-ML/algorithms/gaussian-process-regression/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import numpy as np

# A few noisy samples of a smooth curve, and a fine grid to predict on.
rng = np.random.default_rng(7)
X_train = rng.uniform(-1, 1, size=(12, 1))
y_train = np.sin(3 * X_train[:, 0]) + rng.normal(0, 0.05, size=12)
X_test = np.linspace(-1, 1, 50).reshape(-1, 1)
```

## Code from the page

```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, WhiteKernel

model = GaussianProcessRegressor(RBF(length_scale=0.3) + WhiteKernel(0.01)).fit(X_train, y_train)
mean, std = model.predict(X_test, return_std=True)
```

## Try it

```python
print(mean[:5], std[:5])  # confident near data, unsure between points
```
