# Polynomial Regression

Capture curved relationships by running linear regression on powers of the input.

This notebook comes from the [Polynomial Regression](https://pabl0aranda.github.io/Atlas-ML/algorithms/polynomial-regression/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import numpy as np

# A noisy curve with one input, which a straight line can't follow.
rng = np.random.default_rng(7)
X = rng.uniform(-1, 1, size=(80, 1))
y = 0.8 * X[:, 0] ** 3 - 0.5 * X[:, 0] + rng.normal(0, 0.1, size=80)
```

## Code from the page

```python
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler

model = make_pipeline(PolynomialFeatures(degree=5), StandardScaler(), Ridge(alpha=1.0))
model.fit(X, y)
```
