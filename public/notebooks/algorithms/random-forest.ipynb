# Random Forest

Average many decorrelated decision trees, each trained on a bootstrap sample, into one stable classifier.

This notebook comes from the [Random Forest](https://pabl0aranda.github.io/Atlas-ML/algorithms/random-forest/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)
```

## Code from the page

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=200, oob_score=True, random_state=0).fit(X, y)
print(model.oob_score_)
```
