# Linear Discriminant Analysis

Find the direction that best separates classes, assuming they share one covariance.

This notebook comes from the [Linear Discriminant Analysis](https://pabl0aranda.github.io/Atlas-ML/algorithms/linear-discriminant-analysis/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)
```

## Code from the page

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

model = LinearDiscriminantAnalysis().fit(X_train, y_train)
X_projected = model.transform(X_train)  # at most (classes - 1) dimensions
```

## Try it

```python
print(X_projected.shape)  # two classes, so one dimension
print(f"Test accuracy: {model.score(X_test, y_test):.1%}")
```
