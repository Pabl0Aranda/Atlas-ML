# Support Vector Machines

Find the maximum-margin boundary between two classes, optionally through a kernel.

This notebook comes from the [Support Vector Machines](https://pabl0aranda.github.io/Atlas-ML/algorithms/svm/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)
```

## Code from the page

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

model = make_pipeline(StandardScaler(), SVC(kernel="rbf", C=1.0, gamma="scale"))
model.fit(X, y)
print(model[-1].n_support_)
```
