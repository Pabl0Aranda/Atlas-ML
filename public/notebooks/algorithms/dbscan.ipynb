# DBSCAN

Grow clusters outward from dense regions, finding any shape and labeling isolated points as noise.

This notebook comes from the [DBSCAN](https://pabl0aranda.github.io/Atlas-ML/algorithms/dbscan/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two clean half-moons: density and linkage methods can follow their shape.
X, _ = make_moons(n_samples=300, noise=0.05, random_state=7)
```

## Code from the page

```python
from sklearn.cluster import DBSCAN

labels = DBSCAN(eps=0.15, min_samples=4).fit_predict(X)  # -1 marks noise
```

## Try it

```python
print("Clusters:", len(set(labels) - {-1}), "| noise points:", (labels == -1).sum())
```
