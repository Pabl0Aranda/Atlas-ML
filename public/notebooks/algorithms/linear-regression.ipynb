# Linear Regression

Fit a straight line to data by minimizing mean squared error with gradient descent.

This notebook comes from the [Linear Regression](https://pabl0aranda.github.io/Atlas-ML/algorithms/linear-regression/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import numpy as np

# A noisy straight-line trend with one input.
rng = np.random.default_rng(7)
X = rng.uniform(-1, 1, size=(100, 1))
y = 0.7 * X[:, 0] + 0.1 + rng.normal(0, 0.1, size=100)
```

## Code from the page

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression().fit(X, y)
print(model.intercept_, model.coef_)
```
