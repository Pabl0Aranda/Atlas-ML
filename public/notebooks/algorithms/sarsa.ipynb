# State-Action-Reward-State-Action

SARSA, the on-policy counterpart of Q-Learning, which learns from the actions it really takes.

This notebook comes from the [State-Action-Reward-State-Action](https://pabl0aranda.github.io/Atlas-ML/algorithms/sarsa/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Code from the page

```python
def sarsa_update(Q, s, a, r, s_next, a_next, done, alpha=0.5, gamma=1.0):
    future = 0.0 if done else Q[s_next][a_next]
    Q[s][a] += alpha * (r + gamma * future - Q[s][a])
```

## Try it

```python
# Two states with two actions each. Taking action 1 from "start" reaches the goal for a reward of -1.
Q = {"start": [0.0, 0.0], "goal": [0.0, 0.0]}
sarsa_update(Q, "start", 1, r=-1, s_next="goal", a_next=0, done=True)
print(Q)  # the value of that action moves halfway to -1
```
