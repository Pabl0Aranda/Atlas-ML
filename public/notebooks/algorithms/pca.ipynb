# Principal Component Analysis

Find the orthogonal directions along which data varies most.

This notebook comes from the [Principal Component Analysis](https://pabl0aranda.github.io/Atlas-ML/algorithms/pca/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import numpy as np

# A stretched, tilted cloud: most of the spread lies along one direction.
rng = np.random.default_rng(7)
X = rng.multivariate_normal([0, 0], [[1, 0.9], [0.9, 1]], size=300)
```

## Code from the page

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

Z = PCA(n_components=2).fit(StandardScaler().fit_transform(X))
print(Z.explained_variance_ratio_)
```
