# K-Means Clustering

Partition data into k clusters by alternating between assigning points and moving centroids.

This notebook comes from the [K-Means Clustering](https://pabl0aranda.github.io/Atlas-ML/algorithms/k-means/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_blobs

# Four round groups of points, like the sandbox's Blobs dataset.
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.6, random_state=7)
```

## Code from the page

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters=4, init="k-means++", n_init=10, random_state=0)
labels = model.fit_predict(X)
print(model.cluster_centers_, model.inertia_)
```
