# Multilayer Perceptron

A feedforward neural network trained with backpropagation.

This notebook comes from the [Multilayer Perceptron](https://pabl0aranda.github.io/Atlas-ML/algorithms/mlp/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import torch
from sklearn.datasets import make_moons

# The Moons data as PyTorch tensors.
X_numpy, y_numpy = make_moons(n_samples=300, noise=0.2, random_state=7)
X = torch.tensor(X_numpy, dtype=torch.float32)
y = torch.tensor(y_numpy, dtype=torch.float32)
```

## Code from the page

```python
import torch
from torch import nn

model = nn.Sequential(nn.Linear(2, 6), nn.Tanh(), nn.Linear(6, 6), nn.Tanh(), nn.Linear(6, 1))
optimizer = torch.optim.SGD(model.parameters(), lr=0.5)
loss_fn = nn.BCEWithLogitsLoss()

for epoch in range(1000):
    optimizer.zero_grad()
    loss = loss_fn(model(X).squeeze(1), y)
    loss.backward()
    optimizer.step()
```

## Try it

```python
with torch.no_grad():
    accuracy = ((model(X).squeeze(1) > 0) == y.bool()).float().mean()
print(f"Training accuracy: {accuracy:.1%}")
```
