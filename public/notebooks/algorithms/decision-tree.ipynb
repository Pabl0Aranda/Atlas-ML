# Decision Trees

Classify data with a hierarchy of axis-aligned questions chosen to reduce impurity.

This notebook comes from the [Decision Trees](https://pabl0aranda.github.io/Atlas-ML/algorithms/decision-tree/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)
```

## Code from the page

```python
from sklearn.tree import DecisionTreeClassifier, export_text

model = DecisionTreeClassifier(criterion="gini", max_depth=4).fit(X, y)
print(export_text(model))
```
