# Lasso Regression

Least squares with an L1 penalty that sets coefficients exactly to zero.

This notebook comes from the [Lasso Regression](https://pabl0aranda.github.io/Atlas-ML/algorithms/lasso-regression/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split

# 20 inputs, only 5 of which matter: room for a penalty to make a difference.
X, y = make_regression(n_samples=200, n_features=20, n_informative=5, noise=10, random_state=7)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

## Code from the page

```python
from sklearn.linear_model import LassoCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

model = make_pipeline(StandardScaler(), LassoCV(cv=5)).fit(X_train, y_train)
print((model[-1].coef_ != 0).sum(), "features kept")
```
