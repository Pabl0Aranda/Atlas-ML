# Naive Bayes

Classify with Bayes' rule, assuming features are independent within each class.

This notebook comes from the [Naive Bayes](https://pabl0aranda.github.io/Atlas-ML/algorithms/naive-bayes/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)
```

## Code from the page

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB().fit(X_train, y_train)
print(model.predict_proba(X_test))
```
