# Gaussian Mixture Models

Soft clustering with a weighted sum of Gaussians, fitted by expectation-maximization.

This notebook comes from the [Gaussian Mixture Models](https://pabl0aranda.github.io/Atlas-ML/algorithms/gaussian-mixture-models/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_blobs

# Four round groups of points, like the sandbox's Blobs dataset.
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.6, random_state=7)
```

## Code from the page

```python
from sklearn.mixture import GaussianMixture

model = GaussianMixture(n_components=4, covariance_type="full", n_init=5).fit(X)
print(model.predict_proba(X), model.bic(X))
```
