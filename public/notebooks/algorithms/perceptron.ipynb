# Perceptron

The original learning neuron, a linear classifier trained by correcting its mistakes.

This notebook comes from the [Perceptron](https://pabl0aranda.github.io/Atlas-ML/algorithms/perceptron/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)
```

## Code from the page

```python
from sklearn.linear_model import Perceptron

model = Perceptron(eta0=0.1, max_iter=1000).fit(X, y)
print(model.coef_, model.intercept_)
```
