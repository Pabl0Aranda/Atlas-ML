# Gradient Boosting

Build a strong classifier one small tree at a time, each one correcting the errors left by the trees before it.

This notebook comes from the [Gradient Boosting](https://pabl0aranda.github.io/Atlas-ML/algorithms/gradient-boosting/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)
```

## Code from the page

```python
from sklearn.ensemble import GradientBoostingClassifier

model = GradientBoostingClassifier(n_estimators=200, learning_rate=0.1, max_depth=2).fit(X, y)
print(model.score(X_test, y_test))
```
