# Hierarchical Clustering

Merge the closest clusters step by step to build a tree of nested groups, then cut it at the level you need.

This notebook comes from the [Hierarchical Clustering](https://pabl0aranda.github.io/Atlas-ML/algorithms/hierarchical-clustering/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two clean half-moons: density and linkage methods can follow their shape.
X, _ = make_moons(n_samples=300, noise=0.05, random_state=7)
```

## Code from the page

```python
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

labels = AgglomerativeClustering(n_clusters=2, linkage="single").fit_predict(X)
dendrogram(linkage(X, method="single"))
```
