# Elastic Net

Blend the L1 and L2 penalties to get sparsity with stability.

This notebook comes from the [Elastic Net](https://pabl0aranda.github.io/Atlas-ML/algorithms/elastic-net/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split

# 20 inputs, only 5 of which matter: room for a penalty to make a difference.
X, y = make_regression(n_samples=200, n_features=20, n_informative=5, noise=10, random_state=7)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

## Code from the page

```python
from sklearn.linear_model import ElasticNetCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

model = make_pipeline(StandardScaler(), ElasticNetCV(l1_ratio=[0.1, 0.5, 0.9], cv=5))
model.fit(X_train, y_train)
```
