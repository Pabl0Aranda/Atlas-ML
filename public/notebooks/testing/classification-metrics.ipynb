# Classification Metrics

Confusion matrix, precision, recall, F1 and ROC AUC, and when to use each.

This notebook comes from the [Classification Metrics](https://pabl0aranda.github.io/Atlas-ML/testing/classification-metrics/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)

from sklearn.linear_model import LogisticRegression

model = LogisticRegression().fit(X_train, y_train)
```

## Code from the page

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

predictions = model.predict(X_test)
print(confusion_matrix(y_test, predictions))
print(classification_report(y_test, predictions))
print(roc_auc_score(y_test, model.predict_proba(X_test)[:, 1]))
```
