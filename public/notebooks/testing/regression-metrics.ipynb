MAE, RMSE, R² and MAPE, and how to read residuals.

This notebook comes from the [Regression Metrics](https://pabl0aranda.github.io/Atlas-ML/testing/regression-metrics/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

X, y = make_regression(n_samples=200, n_features=3, noise=15, random_state=7)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
model = LinearRegression().fit(X_train, y_train)
```

## Code from the page

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

predictions = model.predict(X_test)
print(mean_absolute_error(y_test, predictions))
print(mean_squared_error(y_test, predictions) ** 0.5)
print(r2_score(y_test, predictions))
```
