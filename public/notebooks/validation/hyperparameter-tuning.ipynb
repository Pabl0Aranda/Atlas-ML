# Hyperparameter Tuning

Use validation scores to choose the settings a model cannot learn by itself.

This notebook comes from the [Hyperparameter Tuning](https://pabl0aranda.github.io/Atlas-ML/validation/hyperparameter-tuning/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)
```

## Code from the page

```python
from scipy.stats import loguniform
from sklearn.model_selection import RandomizedSearchCV
from sklearn.svm import SVC

search = RandomizedSearchCV(
    SVC(),
    {"C": loguniform(1e-2, 1e2), "gamma": loguniform(1e-3, 1e1)},
    n_iter=30,
    cv=5,
    random_state=0,
)
search.fit(X_train, y_train)
print(search.best_params_, search.best_score_)
```
