## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# A model trained on part of the data, with predictions on held-out validation data.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.3, stratify=y, random_state=0)
model = LogisticRegression().fit(X_train, y_train)
predictions = model.predict(X_val)
probabilities = model.predict_proba(X_val)
```

## Code from the page

```python
from sklearn.metrics import (
    balanced_accuracy_score, cohen_kappa_score, fbeta_score, log_loss,
    matthews_corrcoef, mean_absolute_error, silhouette_score,
)

print(balanced_accuracy_score(y_val, predictions))
print(matthews_corrcoef(y_val, predictions))
print(cohen_kappa_score(y_val, predictions))
print(fbeta_score(y_val, predictions, beta=2))
print(log_loss(y_val, probabilities))
```
