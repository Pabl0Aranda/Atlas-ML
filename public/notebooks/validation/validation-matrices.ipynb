# Validation Matrices

The confusion matrix and the other grids used to judge models, with their uses, how to read them, and why they matter.

This notebook comes from the [Validation Matrices](https://pabl0aranda.github.io/Atlas-ML/validation/validation-matrices/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# A model trained on part of the data, with predictions on held-out validation data.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.3, stratify=y, random_state=0)
model = LogisticRegression().fit(X_train, y_train)
predictions = model.predict(X_val)
probabilities = model.predict_proba(X_val)
```

## Code from the page

```python
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
from sklearn.model_selection import cross_val_score
from statsmodels.stats.contingency_tables import mcnemar

print(confusion_matrix(y_val, predictions))
print(confusion_matrix(y_val, predictions, normalize="true"))  # by row
ConfusionMatrixDisplay.from_predictions(y_val, predictions)

fold_scores = cross_val_score(model, X, y, cv=5)  # one row of a fold score matrix

print(mcnemar([[180, 12], [4, 4]], exact=False, correction=True))
```
