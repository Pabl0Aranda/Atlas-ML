# Data Cleaning

Handle missing values, duplicates, inconsistent entries and outliers.

This notebook comes from the [Data Cleaning](https://pabl0aranda.github.io/Atlas-ML/data-preparation/cleaning/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import pandas as pd

# A small table with the usual problems: a duplicate row, inconsistent spelling,
# a missing value and an outlier price.
df = pd.DataFrame({
    "country": [" Spain", "spain", "France ", "france", "Spain", " Spain", "Germany", None],
    "price": [12.5, 13.0, 11.0, 12.0, 250.0, 12.5, None, 14.0],
})
```

## Code from the page

```python
df = df.drop_duplicates()
df["country"] = df["country"].str.strip().str.lower()
print(df.isna().mean())  # share of missing values per column

q1, q3 = df["price"].quantile([0.25, 0.75])
iqr = q3 - q1
outliers = df[(df["price"] < q1 - 1.5 * iqr) | (df["price"] > q3 + 1.5 * iqr)]
```
