Scale numeric features, encode categories and build features algorithms can use.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
import numpy as np

# Two features on very different scales, and a column of color names.
rng = np.random.default_rng(7)
X_train = rng.normal([50, 0.5], [20, 0.1], size=(200, 2))
X_test = rng.normal([50, 0.5], [20, 0.1], size=(50, 2))
colors_train = [["red"], ["green"], ["blue"], ["green"]]
```

## Code from the page

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler

scaler = StandardScaler().fit(X_train)   # learn from training data only
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

encoder = OneHotEncoder(handle_unknown="ignore").fit(colors_train)
```

## Try it

```python
print(X_train_scaled.mean(axis=0).round(2), X_train_scaled.std(axis=0).round(2))
print(encoder.transform([["blue"], ["purple"]]).toarray())  # an unseen color becomes all zeros
```
