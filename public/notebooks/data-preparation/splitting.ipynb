# Data Splitting

Separate training, validation and test data so you can measure performance honestly.

This notebook comes from the [Data Splitting](https://pabl0aranda.github.io/Atlas-ML/data-preparation/splitting/) page on Atlas ML, an interactive guide to machine learning. Run the cells from top to bottom.

## Example data

The code on the page assumes some data already exists. This cell creates it.

```python
from sklearn.datasets import make_moons

# Two interleaving half-moons, like the sandbox's Moons dataset.
X, y = make_moons(n_samples=300, noise=0.25, random_state=7)

from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=5)
```

## Code from the page

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=0)

folds = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
print(cross_val_score(model, X_train, y_train, cv=folds).mean())
```
